```python
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b
```

```python
result_print = add_print(3, 4)
print("Ergebnis von add_print:", result_print)
```

```python
result_return = add_return(3, 4)
print("Ergebnis von add_return:", result_return)
```

```python
print("Multipliziert (print-Version):", result_print * 2)
print("Multipliziert (return-Version):", result_return * 2)
```

```python
print("Multipliziert (return-Version):", result_return * 2)
```
